## 1. Configuración
Cambie solamente esta celda para preparar una prueba. Use `MODO = 'manual'` para fijar todos los parámetros o `MODO = 'defensa_aleatoria'` para sortear un caso reproducible.

```python
MODO = 'manual'                 # 'manual' o 'defensa_aleatoria'
SEMILLA = 20260918              # guardar para reproducir exactamente la prueba
NOMBRE_PRUEBA = 'defensa_grupo_01'
FS = 16_000                     # Hz
DURACION = 6.0                  # s
PICO_MAXIMO = 0.80              # amplitud digital final
FADE_MS = 20                    # entrada/salida suave

# Señal útil: 'tono', 'multitono', 'chirp', 'armonicos' o 'transitoria'
TIPO_SENAL = 'multitono'
FRECUENCIAS_UTILES = [440, 880, 1320]
AMPLITUDES_UTILES = [1.0, 0.55, 0.30]
FASES_UTILES_GRADOS = [0, 35, -50]

# Ruido: 'tonal', 'red_60hz', 'blanco', 'banda', 'impulsivo' o 'mixto'
TIPO_RUIDO = 'mixto'
SNR_OBJETIVO_DB = 2.0
FRECUENCIA_INTERFERENCIA = 1750  # Hz, para tonal o mixto
BANDA_RUIDO = (2500, 3800)       # Hz, debe quedar bajo FS/2
NUM_IMPULSOS = 10

CARPETA_SALIDA = Path('/kaggle/working/tonos_defensa') if Path('/kaggle/working').exists() else Path('tonos_defensa')
CARPETA_SALIDA
```

## 2. Funciones de generación y medición

```python
def rms(x):
    return float(np.sqrt(np.mean(np.square(x, dtype=np.float64))))

def snr_db(clean, observed):
    error = observed - clean
    return 10*np.log10((np.sum(clean**2)+1e-15)/(np.sum(error**2)+1e-15))

def fade(x, fs, ms=20):
    n = min(int(fs*ms/1000), len(x)//2)
    if n > 0:
        ramp = np.sin(np.linspace(0, np.pi/2, n))**2
        x = x.copy(); x[:n] *= ramp; x[-n:] *= ramp[::-1]
    return x

def normalizar_conjunto(*signals, peak=0.8):
    maximum = max(np.max(np.abs(x)) for x in signals)
    gain = peak/maximum if maximum > 0 else 1.0
    return [x*gain for x in signals], gain

def generar_limpia(kind, t, freqs, amps, phases_deg, rng):
    if kind == 'tono':
        return amps[0]*np.sin(2*np.pi*freqs[0]*t + np.deg2rad(phases_deg[0]))
    if kind == 'multitono':
        return sum(a*np.sin(2*np.pi*f*t + np.deg2rad(p)) for f,a,p in zip(freqs,amps,phases_deg))
    if kind == 'chirp':
        return chirp(t, f0=freqs[0], f1=freqs[-1], t1=t[-1], method='linear')
    if kind == 'armonicos':
        f0 = freqs[0]
        return sum((1/k)*np.sin(2*np.pi*k*f0*t + rng.uniform(-np.pi,np.pi)) for k in range(1,6))
    if kind == 'transitoria':
        env = np.exp(-5*(t % 1.0))*(np.sin(2*np.pi*2*t)>0)
        return env*np.sin(2*np.pi*freqs[0]*t)
    raise ValueError(f'Tipo de señal no reconocido: {kind}')

def ruido_banda(n, fs, band, rng):
    low, high = band
    if not (0 < low < high < fs/2):
        raise ValueError(f'BANDA_RUIDO debe cumplir 0 < low < high < {fs/2}')
    sos = butter(6, [low, high], btype='bandpass', fs=fs, output='sos')
    return sosfiltfilt(sos, rng.standard_normal(n))

def generar_ruido(kind, t, fs, f_int, band, impulses, rng):
    n = len(t)
    tonal = np.sin(2*np.pi*f_int*t + rng.uniform(-np.pi,np.pi))
    red = sum((1/k)*np.sin(2*np.pi*60*k*t + rng.uniform(-np.pi,np.pi)) for k in range(1,5))
    white = rng.standard_normal(n)
    band_noise = ruido_banda(n, fs, band, rng)
    impulse = np.zeros(n)
    locations = rng.choice(np.arange(int(.1*fs), n-int(.1*fs)), size=min(impulses,max(1,n//100)), replace=False)
    for loc in locations:
        width = max(2, int(rng.uniform(.0005,.004)*fs))
        stop = min(n, loc+width)
        impulse[loc:stop] += rng.choice([-1,1])*np.hanning(2*(stop-loc))[:stop-loc]
    options = {'tonal':tonal, 'red_60hz':red, 'blanco':white, 'banda':band_noise, 'impulsivo':impulse}
    if kind == 'mixto':
        return 0.55*tonal + 0.30*band_noise + 0.15*impulse
    if kind not in options: raise ValueError(f'Tipo de ruido no reconocido: {kind}')
    return options[kind]

def escalar_ruido_para_snr(clean, noise, target_db):
    target_noise_rms = rms(clean)/(10**(target_db/20))
    return noise*(target_noise_rms/(rms(noise)+1e-15))

def espectro(x, fs):
    w = get_window('hann', len(x), fftbins=True)
    X = np.fft.rfft(x*w)
    f = np.fft.rfftfreq(len(x), 1/fs)
    mag = 20*np.log10(np.maximum(np.abs(X)/(np.sum(w)/2), 1e-12))
    return f, mag
```

## 3. Sorteo opcional de la prueba desconocida
El sorteo evita casos imposibles para el hardware: todas las frecuencias permanecen por debajo de Nyquist y la banda de ruido se construye dentro del margen disponible. La configuración final siempre se muestra y se guarda.

```python
rng = np.random.default_rng(SEMILLA)
if MODO == 'defensa_aleatoria':
    TIPO_SENAL = rng.choice(['tono','multitono','armonicos','transitoria']).item()
    TIPO_RUIDO = rng.choice(['tonal','red_60hz','blanco','banda','impulsivo','mixto']).item()
    base = int(rng.choice([220, 260, 330, 390, 440, 520, 660]))
    FRECUENCIAS_UTILES = [base, min(2*base, int(.28*FS)), min(3*base, int(.38*FS))]
    AMPLITUDES_UTILES = [1.0, float(rng.uniform(.35,.7)), float(rng.uniform(.15,.4))]
    FASES_UTILES_GRADOS = rng.integers(-180,181,3).tolist()
    FRECUENCIA_INTERFERENCIA = int(rng.uniform(.18*FS, .40*FS))
    low = int(rng.uniform(.20*FS,.30*FS)); high = int(min(low+rng.uniform(.07*FS,.14*FS),.46*FS))
    BANDA_RUIDO = (low, high)
    SNR_OBJETIVO_DB = float(rng.choice([-5,-2,0,3,6]))

print({'modo':MODO,'semilla':SEMILLA,'señal':TIPO_SENAL,'ruido':TIPO_RUIDO,
       'frecuencias_utiles_Hz':FRECUENCIAS_UTILES,'interferencia_Hz':FRECUENCIA_INTERFERENCIA,
       'banda_ruido_Hz':BANDA_RUIDO,'snr_objetivo_dB':SNR_OBJETIVO_DB})
```

## 4. Generación, comprobación y gráficas

```python
if FS < 2*max(FRECUENCIAS_UTILES + [FRECUENCIA_INTERFERENCIA, BANDA_RUIDO[1]]):
    raise ValueError('La configuración viola Nyquist. Aumente FS o reduzca las frecuencias.')

t = np.arange(int(FS*DURACION))/FS
clean = generar_limpia(TIPO_SENAL,t,FRECUENCIAS_UTILES,AMPLITUDES_UTILES,FASES_UTILES_GRADOS,rng)
clean = fade(clean,FS,FADE_MS)
noise_raw = generar_ruido(TIPO_RUIDO,t,FS,FRECUENCIA_INTERFERENCIA,BANDA_RUIDO,NUM_IMPULSOS,rng)
noise = escalar_ruido_para_snr(clean,noise_raw,SNR_OBJETIVO_DB)
noisy = clean + noise
[clean, noise, noisy], gain = normalizar_conjunto(clean,noise,noisy,peak=PICO_MAXIMO)
snr_real = snr_db(clean,noisy)
print(f'Muestras: {len(t):,} | SNR obtenida: {snr_real:.3f} dB | Ganancia común: {gain:.5f}')

view = min(len(t),int(.05*FS))
f1,m1 = espectro(clean,FS); f2,m2 = espectro(noisy,FS)
fig,ax = plt.subplots(2,1,figsize=(14,8))
ax[0].plot(t[:view]*1000,clean[:view],label='Limpia',lw=1.5)
ax[0].plot(t[:view]*1000,noisy[:view],label='Contaminada',alpha=.7,lw=1)
ax[0].set(xlabel='Tiempo (ms)',ylabel='Amplitud',title='Primeros 50 ms'); ax[0].legend()
ax[1].plot(f1,m1,label='Limpia'); ax[1].plot(f2,m2,label='Contaminada',alpha=.75)
ax[1].set_xlim(0,FS/2); ax[1].set_ylim(max(-120,np.max(m2)-100),np.max(m2)+5)
ax[1].set(xlabel='Frecuencia (Hz)',ylabel='Magnitud (dB)',title='Espectro con ventana Hann'); ax[1].legend()
plt.tight_layout(); plt.show()
```

## 7.1 DFT directa y FFT


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Ejemplo numérico: DFT directa vs FFT
# ============================================================

x = np.array([1.0, 0.0, -1.0, 0.0])
N = len(x)

def dft_directa(x):
    N = len(x)
    X = np.zeros(N, dtype=complex)

    for k in range(N):
        for n in range(N):
            X[k] += x[n] * np.exp(
                -1j * 2 * np.pi * k * n / N
            )

    return X

X_dft = dft_directa(x)
X_fft = np.fft.fft(x)

print("Señal x[n]:")
print(x)

print("\nDFT directa:")
print(np.round(X_dft, 6))

print("\nFFT:")
print(np.round(X_fft, 6))

print("\nDiferencia máxima:")
print(np.max(np.abs(X_dft - X_fft)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Magnitud y fase del ejemplo
# ============================================================

magnitud = np.abs(X_dft)
fase = np.angle(X_dft)

# Si la magnitud es prácticamente cero,
# la fase no tiene significado físico útil.
fase[magnitud < 1e-10] = 0.0

k = np.arange(N)

fig, ax = plt.subplots(
    2,
    1,
    figsize=(8, 6)
)

ax[0].stem(
    k,
    magnitud
)

ax[0].set_title(
    "Magnitud de la DFT"
)

ax[0].set_xlabel(
    "Bin k"
)

ax[0].set_ylabel(
    "|X[k]|"
)

ax[0].grid(True)


ax[1].stem(
    k,
    fase
)

ax[1].set_title(
    "Fase de la DFT"
)

ax[1].set_xlabel(
    "Bin k"
)

ax[1].set_ylabel(
    "Fase (rad)"
)

ax[1].grid(True)


plt.tight_layout()

ruta = (
    "/kaggle/working/"
    "dft_ejemplo_magnitud_fase.png"
)

plt.savefig(
    ruta,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "Figura guardada en:",
    ruta
)
```

## 7.2 Resolución frecuencial, latencia y memoria


```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Resolución frecuencial, duración de bloque y memoria
# ============================================================

fs = 16000  # Hz

tamanos_bloque = np.array([
    128,
    256,
    512,
    1024,
    2048,
    4096
])

duraciones = tamanos_bloque / fs

delta_f = fs / tamanos_bloque

# Memoria aproximada para un bloque real float32
bytes_por_muestra = 4

memoria_bytes = (
    tamanos_bloque *
    bytes_por_muestra
)

memoria_kb = (
    memoria_bytes / 1024
)

print(
    "fs =",
    fs,
    "Hz"
)

print()
print(
    "N\tT (ms)\tΔf (Hz)\tMemoria float32 (kB)"
)

for N, T, df, mem in zip(
    tamanos_bloque,
    duraciones,
    delta_f,
    memoria_kb
):
    print(
        f"{N}\t"
        f"{T*1000:.2f}\t"
        f"{df:.3f}\t"
        f"{mem:.2f}"
    )
```

```python
fig, ax1 = plt.subplots(
    figsize=(8, 5)
)

ax1.plot(
    tamanos_bloque,
    delta_f,
    marker="o"
)

ax1.set_xlabel(
    "Tamaño de bloque N"
)

ax1.set_ylabel(
    "Resolución frecuencial Δf (Hz)"
)

ax1.set_title(
    "Compromiso entre tamaño de bloque y resolución frecuencial"
)

ax1.grid(True)

plt.tight_layout()

ruta = (
    "/kaggle/working/"
    "resolucion_frecuencial.png"
)

plt.savefig(
    ruta,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "Figura guardada en:",
    ruta
)
```

# 8. Implementación de los métodos de filtrado

Se implementan dos métodos de reducción de ruido: un filtro temporal IIR y un filtro espectral mediante FFT/IFFT.

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile

from src.filtros_temporales import (
    disenar_filtro,
    ecuacion_diferencias,
    propiedades,
    validar,
)

from src.filtrado_espectral import filtrar_espectral

FS = 16000
N = 1024

BASE = Path.cwd()
PRUEBAS = BASE / "pruebas"

def leer_wav(ruta):
    fs, datos = wavfile.read(ruta)

    if fs != FS:
        raise ValueError(f"Frecuencia de muestreo inesperada: {fs}")

    if datos.ndim > 1:
        raise ValueError("Se esperaba una señal mono.")

    if np.issubdtype(datos.dtype, np.integer):
        escala = max(
            abs(np.iinfo(datos.dtype).min),
            np.iinfo(datos.dtype).max
        )
        datos = datos.astype(np.float64) / escala
    else:
        datos = datos.astype(np.float64)

    return datos


def localizar_wav(carpeta, prefijo):
    archivos = sorted(carpeta.glob(f"{prefijo}*.wav"))

    if len(archivos) != 1:
        raise FileNotFoundError(
            f"Se esperaba un WAV con prefijo {prefijo} en {carpeta}. "
            f"Encontrados: {len(archivos)}"
        )

    return archivos[0]


casos = {}

for numero in (1, 2, 3):
    carpeta = PRUEBAS / f"caso_{numero}"

    limpia = leer_wav(localizar_wav(carpeta, "01_"))
    ruido = leer_wav(localizar_wav(carpeta, "02_"))
    contaminada = leer_wav(localizar_wav(carpeta, "03_"))

    if not (len(limpia) == len(ruido) == len(contaminada)):
        raise ValueError(f"Longitudes incompatibles en caso {numero}")

    casos[numero] = {
        "limpia": limpia,
        "ruido": ruido,
        "contaminada": contaminada,
    }

print("Tres casos cargados correctamente.")
```

## 8.2 Aplicación del filtro temporal

Se aplica el filtro IIR a los tres casos de prueba y se comparan las señales originales, contaminadas y filtradas.

```python
# Aplicación del filtrado espectral a los tres casos

detalles_espectrales = {}

for numero, datos in casos.items():

    salida, detalles = filtrar_espectral(
        datos["contaminada"],
        fs=FS,
        N=N,
        reduccion=1.0,
        detalle=True
    )

    datos["espectral"] = salida
    detalles_espectrales[numero] = detalles

    print(
        f"Caso {numero}: "
        f"{len(salida)} muestras, "
        f"interferencias detectadas: "
        f"{detalles['frecuencias_picos_hz']}"
    )
```

## 8.3 Comparación de los métodos de filtrado

Se comparan los resultados del filtro temporal IIR y del método espectral FFT/IFFT mediante el cálculo del MSE, SNR de entrada, SNR de salida y mejora de SNR.

```python
import pandas as pd


def calcular_mse(referencia, estimacion):
    referencia = np.asarray(referencia, dtype=np.float64)
    estimacion = np.asarray(estimacion, dtype=np.float64)

    return float(np.mean((referencia - estimacion) ** 2))


def calcular_snr(referencia, estimacion):
    referencia = np.asarray(referencia, dtype=np.float64)
    estimacion = np.asarray(estimacion, dtype=np.float64)

    energia_senal = np.sum(referencia ** 2)
    energia_error = np.sum((referencia - estimacion) ** 2)

    if energia_error == 0:
        return float("inf")

    return float(
        10 * np.log10(energia_senal / energia_error)
    )


registros = []

for numero, datos in casos.items():

    referencia = datos["limpia"]
    entrada = datos["contaminada"]

    snr_entrada = calcular_snr(referencia, entrada)

    for metodo in ("temporal", "espectral"):

        estimacion = datos[metodo]
        snr_salida = calcular_snr(referencia, estimacion)

        registros.append({
            "Caso": numero,
            "Metodo": metodo,
            "MSE": calcular_mse(referencia, estimacion),
            "SNR entrada (dB)": snr_entrada,
            "SNR salida (dB)": snr_salida,
            "Mejora SNR (dB)": snr_salida - snr_entrada,
        })

tabla_resultados = pd.DataFrame(registros)

display(tabla_resultados.round(4))
```

```python
# Evaluación de diferentes intensidades de reducción

reducciones = [0.0, 0.5, 1.0, 1.5, 2.0, 3.0]

referencia = casos[3]["limpia"]
contaminada = casos[3]["contaminada"]

snr_obtenidas = []
mse_obtenidos = []

for reduccion in reducciones:

    salida = filtrar_espectral(
        contaminada,
        fs=FS,
        N=N,
        reduccion=reduccion
    )

    snr_obtenidas.append(
        calcular_snr(referencia, salida)
    )

    mse_obtenidos.append(
        calcular_mse(referencia, salida)
    )

fig, ejes = plt.subplots(1, 2, figsize=(12, 4))

ejes[0].plot(reducciones, snr_obtenidas, marker="o")
ejes[0].set_title("Caso 3: SNR vs. reducción")
ejes[0].set_xlabel("Parámetro de reducción")
ejes[0].set_ylabel("SNR (dB)")
ejes[0].grid(True)

ejes[1].plot(reducciones, mse_obtenidos, marker="o")
ejes[1].set_title("Caso 3: MSE vs. reducción")
ejes[1].set_xlabel("Parámetro de reducción")
ejes[1].set_ylabel("MSE")
ejes[1].grid(True)

plt.tight_layout()
plt.show()
```
